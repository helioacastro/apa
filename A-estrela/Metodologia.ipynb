Mapas de jogos (Game Maps) são frequentemente utilizados em pesquisa acadêmica de Inteligência Artificial para Pathfinding (busca de caminhos).

O repositório Moving AI Lab (da Universidade de Denver) mantém uma coleção enorme de mapas de jogos como Dragon Age: Origins, Warcraft III e StarCraft, usados justamente para benchmarks de algoritmos.

O dataset utilizado para realizar os benchmarks foi extraídos do mapa de jogo do Dragon Age: Origins, no endereço https://www.movingai.com/benchmarks/dao/index.html. [https://www.movingai.com/benchmarks/dao/brc202d.map.zip]
